## Python Data Types

Chris Bruehl covers the fundamental Python data types: strings, integers, floats, booleans, lists, tuples, dictionaries, and sets. Below are demonstrations of the  function and type conversion.

```python
# From Bruehl - string type
x = "hello"
print(x, type(x))
```

```python
# From Bruehl - integer type
y = 10
print(y, type(y))
```

```python
# From Bruehl - float type
z = 3.14
print(z, type(z))
```

```python
# From Bruehl - boolean type
b = True
print(b, type(b))
```

```python
# From Bruehl - list type
lst = [1, 2, 3]
print(lst, type(lst))
```

```python
# From Bruehl - tuple type
t = (1, 2)
print(t, type(t))
```

```python
# From Bruehl - dict type
d = {"key": "value"}
print(d, type(d))
```

```python
# My addition - set type
s = {1, 2, 3}
print(s, type(s))
```

```python
# My addition - convert string to int
age_str = "30"
age_int = int(age_str)
print(age_int, type(age_int))
```

```python
# My addition - convert float to int
pi = 3.99
pi_int = int(pi)
print(pi_int, type(pi_int))
```

```python
# Extra credit - multi-step conversion
float_str = "2.5"
result = int(float(float_str))
print(result, type(result))
```

```python
# Extra credit - convert list to tuple
my_list = [1, 2, 3]
my_tuple = tuple(my_list)
print(my_tuple, type(my_tuple))
```

```python
# Extra credit - bool conversions
print(bool(1), type(bool(1)))
print(bool(0), type(bool(0)))
```

## Variables

### Variable Assignment

```python
# From Bruehl - basic assignment
course = "DSCI-6883"
print(course)
```

```python
# From Bruehl - multiple assignment in one line
a, b, c = 1, 2, 3
print(a, b, c)
```

### Overwriting and Deleting Variables

```python
# From Bruehl - overwrite variable
score = 85
print("first:", score)
score = 92
print("after overwrite:", score)
```

```python
# From Bruehl - delete variable
temp = "temporary"
print(temp)
del temp
# print(temp)  # raises NameError
```

### Naming and Tracking Variables

```python
# From Bruehl - meaningful names
student_count = 30
average_score = 87.5
print(f"{student_count} students, average {average_score}")
```

```python
# From Bruehl - track with type()
value = 42
print(value, type(value))
value = "forty-two"
print(value, type(value))
```

### Variable Naming Rules

```python
# From Bruehl - valid snake_case
total_price = 19.99
item_count = 5
_system_note = "internal"
print(total_price, item_count, _system_note)
```

```python
# My addition - invalid names would raise SyntaxError:
# 2var = 5        # starts with number
# my-var = 5      # hyphen not allowed
# class = 5       # reserved keyword
print("see comments above for invalid examples")
```

```python
# My addition - case sensitivity matters
name = "Scott"
Name = "Different"
print(name, Name)
```

```python
# My addition - same value to many variables
x = y = z = 0
print(x, y, z)
```

```python
# My addition - dynamic typing
var = 10
print(type(var))
var = "text"
print(type(var))
```

```python
# My addition - id() shows memory address
x = 500
y = 500
print(id(x), id(y))
```
