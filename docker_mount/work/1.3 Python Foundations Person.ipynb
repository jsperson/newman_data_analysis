# 29. DEMO: Overwriting & Deleting Variables

```python
# Change the value of a variable - last assignment wins
snowboard_inventory = 100
snowboard_inventory = 85
print(snowboard_inventory)
```

```python
snowboard_inventory = 100
print(snowboard_inventory)
```

```python
snowboard_inventory = 85
print(snowboard_inventory)
```

```python
snowboard_inventory = 70
print(snowboard_inventory)
```

```python
# Type 3 SCD (ish)
old_snowboard_inventory = 85
snowboard_inventory = 70
print(old_snowboard_inventory, snowboard_inventory)
```

```python
# Overwrite one variable with another
old_snowboard_inventory = 85
snowboard_inventory = 70

snowboard_inventory = old_snowboard_inventory

print(snowboard_inventory)
```

```python
# Overwrite one variable with another - it's assigned by value not reference
old_snowboard_inventory = 85
snowboard_inventory = 70

snowboard_inventory = old_snowboard_inventory
old_snowboard_inventory = 100

print(snowboard_inventory)
```

```python
# Deleting variables
del old_snowboard_inventory
```

```python
print(old_snowboard_inventory)
```

```python
# Reassign variable and ... it's back.
old_snowboard_inventory = 100
print(old_snowboard_inventory)
```
