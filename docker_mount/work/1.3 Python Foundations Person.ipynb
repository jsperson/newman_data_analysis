## Python Data Types

### Single Value Data Types
* Numeric - numeric values
    * Integer(int)
    * Float(float)
    * Complex(complex) (not commonly used)
* Text - sequence of characters
    * String(str)
* Boolean - True or False value
    * Boolean(bool)
* None - represents the absense of a value
    * NoneType

### Multiple Values
* Sequence - sequence of values, usually text or numeric
    * List(list) - [1,2,3,4]
    * Tuple(tuple) - ('yes','no')
    * Range(range) - range(1,100,1)
* Mapping - maps keys to values
    * Dictionary(dict) - {'foo': 10,'fee': 15}
* Set - collection of unique values
    * Set(set) - {'Scott', 'Jen'}
    * Frozen Set(frozenset) - {'Scott', 'Jen'}

### Variable Naming Rules
Python variable names must follow these rules:

* May contain letters (a-z, A-Z), digits (0-9), and underscores (_)
* Cannot start with a digit (e.g., `2var` is invalid)
* Cannot contain spaces or special characters like hyphens (e.g., `my-var` is invalid)
* Cannot be a reserved Python keyword (e.g., `class`, `if`, `for`)

Names are case-sensitive, so `name` and `Name` are two different variables.
**Best practice (PEP 8):** use `snake_case` — lowercase words separated by underscores, like `total_price` or `student_count`. Pick descriptive names that tell you what the variable holds, e.g., `average_score` instead of `x`.
