# Import Required Modules

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Load the Data

```python
data_path = '/Workspace/Users/jsperson@gmail.com/newman_data_analysis/data/'
```

```python
df = pd.read_csv(data_path + 'loan3000.csv')
```

```python
df = df.rename(columns={'Unnamed: 0': 'borrower_id', 'purpose_': 'purpose'})
```

```python
df
```

## Data Explanation
This data represents a set of 3000 loans. The loans are of various types and appear to have a status of either default or paid off. I suspect this dataset can be used to analyze and predict outcomes based on the attributes like debt to income ratio, borrower (credit) score, and payment to income ratio.  

# Create Four Plots

## Histogram

```python
float_cols = ['dti','borrower_score','payment_inc_ratio']
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, len(float_cols), figsize=(15, 4))
for ax, col in zip(axes, float_cols):
    df[col].hist(ax=ax, bins=30, edgecolor='black')
    ax.set_title(col)
    ax.set_xlabel(col)
    ax.set_ylabel('Count')
plt.tight_layout()
plt.show()
```

### Histogram Plot Discussion
One of the first plots that I like to do is a histogram of any numeric values. This just gives me a look at the distribution of the data to make sure that nothing crazy is going on. We can also see the distribution of the data and start getting a picture of how the data works.

## Box Plot

```python
fig, axes = plt.subplots(1, len(float_cols), figsize=(15, 4))
for ax, col in zip(axes, float_cols):
    df.boxplot(column=col, by='outcome', ax=ax)
    ax.set_title(col)
    ax.set_xlabel('outcome')
    ax.set_ylabel(col)
plt.suptitle('')
plt.tight_layout()
plt.show()
```

### Boxplot Discussion
It's apparent that the purpose of this dataset is to predict outcome based on continuous and cetegorical features. Here I plotted each of the numeric features against outcome using box plots. We start to see the relationship of the features to the outcome and they are as expected. Of particular note is a high number of outliers for payment income ratio for paid off loans. This is interesting and I suspect might be explained by values in the purpose feature.  

## Stacked Bar

```python
purpose_outcome = df.groupby(['purpose', 'outcome']).size().unstack(fill_value=0)
purpose_outcome.plot(kind='bar', stacked=True, figsize=(10, 5), edgecolor='black')
plt.title('Loan Outcome by Purpose')
plt.xlabel('Purpose')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Outcome')
plt.tight_layout()
plt.show()
```

### Stacked Bar Discussion
This provides a visual representation of the number (count) of loans vs. the purpose with a breakdown by outcome. We can see both the relative proportion of loans made for each purpose and get a rough idea as to their relative outcomes. It's not a great graph but we're just at the beginning of EDA in this course... One thing that I can see is that small business loans appear to be significantly more likely to default (a lot of blue relative to the orange on that bar). Similarly credit card loans are the reverse. This would be something I would follow up on with some numeric statistics.

Note this would be really bad as a line chart because then we'd expect to infer trends based on line slope. Best to generally leave line charts for time series data.

## Correlation Heatmap

```python
corr = df[float_cols].corr()
fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, square=True, ax=ax)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

### Correlation Heatmap Discussion
With the correclation heatmap we're looking for numeric variables that...correlate. A strong correlation - around an absolute value of 0.5 or above would indicate features that would contribute duplicative information when predicting an outcome. We'd need to consider removing one of the correlating pairs in order to improve our modeling. In this case we're looking pretty good.
