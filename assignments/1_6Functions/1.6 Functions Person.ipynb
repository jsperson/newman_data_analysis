## Assignment: Defining a Function

### Tax Calculator

Write a function that takes a subtotal and a tax rate and returns the total invoice amount.

Tax is equal to subtotal multiplied by the tax rate.
The total is equal to subtotal plus tax.

````python
def tax_calculator(subtotal, tax_rate):
    """calculates tax amount

    Args:
        subtotal (float): subtotal of order
        tax_rate (float): tax rate

    Returns:
        float: total amount (subtotal + tax)
    """
    tax_amount = subtotal + (subtotal * tax_rate)
    return tax_amount

````

````python
tax_calculator(100, 0.06)
````

## Argument Types
There are several types of arguments that can be passed to a function:
* Positional - identified by the order passed
* Keyword - passed in any order by using the arguments name
* Default - function is defined with default values that are used if nothing is passed
* \*args - arguments pass any number of positional arguments as tuples
* \*kwargs - arguments pass any number of keyword arguments as dictionaries

````python
# Positional
print(concatenator('hello', 'world'))
# Keyword
print(concatenator(string2='world', string1='hello'))
````

Note: positional arguments have to come first then followed by keyword arguments. This is generally done when you don't want to specify all of the positonal arguments.

````python
concatenator('hello', string2='world')
````

````python
# default - set at function creation

def concatenator(string1, string2='world'):
    return string1 + ' ' + string2

concatenator('hola')
````

Note: arguments with defaults must come after arguments without defaults

````python
def concatenator(*args):
    new_string = ''
    for arg in args:
        new_string += (arg + ' ')
    return new_string.rstrip()

concatenator('hello', 'world', 'how', 'are', 'you')
````

Note: *args can just as easily be *words. It's the asterisk that matters not the argument name.

````python
def concatenator(**kwargs): # Note: could be any argument name as long as starts with **
    new_string = ''
    for key, value in kwargs.items():
        new_string += (value + ' ')
    return new_string.rstrip()   
````

````python
concatenator(a='Hello',b='there!',
             c="What's", d='up?')
````

````python
def exponentiator(constant, base, exponent):
    return constant * (base**exponent)
````

````python
# ** unpacks a dictionary
param_dict = {'constant': 2, 'base': 3, 'exponent': 2}
exponentiator(**param_dict)
````

## Return Values

````python
def concatenator(*words):
    sentence = ''
    for word in words:
        sentence += word + ' '
    last_word = words[-1]
    return sentence.rstrip(), last_word
````

````python
sentence, last_word = concatenator('Hello','world!','How','are','you?')

print(sentence)
print(last_word)
````

````python
# Return as a list
def concatenator(*words):
    sentence = ''
    for word in words:
        sentence += word + ' '
    last_word = words[-1]
    return [sentence.rstrip(), last_word]
````

````python
concatenator('Hello','world!','How','are','you?')
````

````python
# Return as a dictionary
def concatenator(*words):
    sentence = ''
    for word in words:
        sentence += word + ' '
    last_word = words[-1]
    return {sentence.rstrip(): last_word}
````

````python
concatenator('Hello','world!','How','are','you?')
````

## Demo: Createing and Importing Modules

````python
%%writefile rating_cleaner.py

def rating_cleaner(rating):
    """gets numeric portion from web rating string
    Args:
        rating (str): rating string
    Returns:
        int: numeric portion of rating
    """
    return int(rating[0])

def rating_list_cleaner(rating_list):
    """cleans lists of ratings and returns numeric ratings lists

    Args:
        rating_list (list): list of rating strings
    Returns:
        list: list of integer ratings
    """
    numeric_list = []
    for rating in rating_list:
        numeric_rating = rating_cleaner(rating)
        numeric_list.append(numeric_rating)
    return numeric_list
````

### Modified Tax Calculator

Since the tax calculator we made was primarily for our Stowe employees, let's modify the tax calculator below to have a default tax rate of 6%.

Also, we want the output of the function to return a list containing the subtotal, tax, and totals.

````python
%%writefile tax_calculator.py
def tax_calculator(subtotal, tax_rate=0.06):
    """calculates price with tax

    Args:
        subtotal (float): price of item
        tax_rate (float): tax rate

    Returns:
        float: price with tax
    """
    tax = round(subtotal*(tax_rate), 2)
    total = round(subtotal + tax,2)
    return[subtotal, tax, total]
````

## Comprehensions
Comprehensions - generate sequences from sequences
```
new_list = [expression for member in other_iterable (if condition)]
```

````python
# for loop method

usd_list = [5.99, 19.99, 24.99, 0, 74.99, 99.99]
exchange_rate = 0.88
euro_list = []

for price in usd_list:
    euro_list.append(round(price * exchange_rate, 2))

euro_list
````

````python
# comprehension

euro_list = [round(price * exchange_rate, 2) for price in usd_list]
euro_list
````
