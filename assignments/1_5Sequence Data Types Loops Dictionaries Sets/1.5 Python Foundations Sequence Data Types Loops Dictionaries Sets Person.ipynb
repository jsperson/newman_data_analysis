## Copying Lists

### Three ways to copy a list
1. Variable assignment - assign a list to a new variable creates a view; the view changes when the base list changes
2. .copy() - applying this method to a list creates a 'shallow' copy; changes to entire elements (nested lists) will not carry over between original and copy; changes to invididual elements within a nested list will still be reflected in both
3. deepcopy() - creates entirely separate lists

````python
list_of_lists = [['a','b','c'],
                 ['d','e','f'],
                 ['g','h','i']]

list_of_lists2 = list_of_lists
list_of_lists[1] = ['x','y','z']

list_of_lists2
````

````python
list_of_lists = [['a','b','c'],
                 ['d','e','f'],
                 ['g','h','i']]

list_of_lists2 = list_of_lists.copy()
list_of_lists[1] = ['x','y','z']

list_of_lists2
````

````python
list_of_lists[0][1] = 'Oh no!'
list_of_lists2
````

````python
from copy import deepcopy

list_of_lists = [['a','b','c'],
                 ['d','e','f'],
                 ['g','h','i']]

list_of_lists2 = deepcopy(list_of_lists)

list_of_lists[0][1] = 'Oh no!'
list_of_lists2

list_of_lists2
````

### Applying Sales Tax

Last weekend our system went down and we didn't have a calculator on site, so we told customers we'd charge them for tax later.

The transactions are below.

````python
subtotals = [15.98, 899.97, 799.97, 117.96, 5.99, 599.99, 24.99, 1799.94, 99.99]
````

Can you apply sales tax to all of the transactions?

The tax rate is 8%.

Once we've calculated the taxes, we'll need to add them to their subtotals and calculate the the total invoice value.

Your code should produce two lists.

* taxes:The first should be a list of the tax amounts on each transaction, rounded to the nearest .01.

* totals: The second should be a list where each element is subtotal[i] + tax[i]

````python
taxes = []
totals = []

for index, tot in enumerate(subtotals):
    tax = round(tot * .08, 2)
    taxes.append(tax)
    total = round(tot + tax, 2)
    totals.append(total)

print(taxes)
print(totals)
````

### Sales Tax by State

We forgot to account for the fact that different states have different tax rates, so we need to modify our loop we just built.

We have a list containing the location of each purchase. The transaction subtotal[i] took place at locations[i].

````python
subtotals = [
    15.98, 899.97, 799.97, 117.96, 5.99, 
    599.99, 24.99, 1799.94, 99.99
]
````

````python
location = [
    'Sun Valley', 'Stowe', 'Mammoth', 'Stowe', 'Sun Valley', 
    'Mammoth', 'Mammoth', 'Mammoth', 'Sun Valley'
]
````

````python
The following tax rates should be applied for each location:

Sun Valley: 8%
Stowe: 6%
Mammoth: 7.75%
````

````python
tax_rates = [0.08, 0.06, 0.0775]
tax_rate_loc = ['Sun Valley','Stowe','Mammoth']

taxes = []
totals = []

for index, tot in enumerate(subtotals):
    cur_loc = location[index]
    cur_loc_index = tax_rate_loc.index(cur_loc)
    tax_rate = tax_rates[cur_loc_index]
    tax = round(tot * tax_rate, 2)
    taxes.append(tax)
    total = round(tot + tax, 2)
    totals.append(total)

print(taxes)
print(totals)
````

### Calculate Projected Inventory for Each Month

We currently have 686 pairs of skis in inventory, and sell 84 each month.

Write a program that keeps track of monthly inventory, by subtracting 84 pairs of skis from inventory each month.

Your program should print 'At the end of month x, we have y pairs of inventory', where x is the number of months that has gone by and y is the remaining inventory at that month.

If inventory goes negative, your program should stop running.

````python
current_inventory = 686
monthly_sales = 84

month_cnt = 0
while current_inventory > 0:
    current_inventory -= monthly_sales
    month_cnt += 1

    print(f'At the end of month {month_cnt}, we have {current_inventory} pairs of skis')
````

## Set Operations
* Union - all unique values present in one or more of both sets
```
set1.union(set2)
```
* Intersection - all unique values present in both sets
```
set1.intersection(set2)
```
* Difference - values present in set 1 but not in set 2
```
set1.difference(set2)
```
* Symetric Difference - returns values not shared between sets
```
set1.symmetric_difference(set2)
```

## Set Use Cases
1. Sets are more efficient than lists for performing membership tests
2. Sets gather unique values without looping
3. Set operations can find the data shared, or not shared, between items without looping
